```python
import sympy as sy
C11, C22, C66, C16, C26, C12 = sy.symbols('C_{11} C_{22} C_{66} C_{16} C_{26} C_{12}')
εxx_1, εyy_1, εxy_1, εxx_2, εyy_2, εxy_2, εxx_3, εyy_3, εxy_3 = sy.symbols('ε_{xx}^{1} ε_{yy}^{1} ε_{xy}^{1} ε_{xx}^{2} ε_{yy}^{2} ε_{xy}^{2} ε_{xx}^{3} ε_{yy}^{3} ε_{xy}^{3}')
sxx_1, syy_1, sxy_1, sxx_2, syy_2, sxy_2, sxx_3, syy_3, sxy_3 = sy.symbols('\sigma_{xx}^{1} \sigma_{yy}^{1} \sigma_{xy}^{1} \sigma_{xx}^{2} \sigma_{yy}^{2} \sigma_{xy}^{2} \sigma_{xx}^{3} \sigma_{yy}^{3} \sigma_{xy}^{3}')
```

```python
esfuerzos = sy.Matrix([[sxx_1, sxx_2, sxx_3],[syy_1, syy_2, syy_3],[sxy_1, sxy_2, sxy_3]])
esfuerzos
```

```python
deformacion = sy.Matrix([[εxx_1, εxx_2, εxx_3],[εyy_1, εyy_2, εyy_3],[2*εxy_1, 2*εxy_2, 2*εxy_3]])
deformacion
```

```python
coeficientes = sy.Matrix([[C11, C12, C16], [C12, C22, C26], [C16, C26, C66]])
coeficientes
```

```python
p1 = esfuerzos * deformacion.T
p2 = deformacion * deformacion.T
```
